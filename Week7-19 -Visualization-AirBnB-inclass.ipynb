Matplotlib is a library for making 2D plots in Python. It is designed with the philosophy that you should be able to create simple plots with just a few commands.

Creating plot consists of 3 steps:

* Step 1: Initialize
* Step 2: Prepare
* Step 3: Render and Observe

1.  Initialize

image.png

```python
import numpy as np
import matplotlib.pyplot as plt
```

2. Prepare

```python
# sales-Six Months
X = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun"])
Y = np.array([100, 150, 130, 200, 170, 180])
```

3. Render & Observe

```python
# Plot

# Create figure
fig, ax = plt.subplots()

# Plot the axes

ax.plot(X, Y)

# Display the plot
fig.show()
```

## Seaborn + Matplotlib
Seaborn can be used in conjunction with Matplotlib. To do so use the Seaborn Plotting function as usual, and then include Matplotlib customisation function.

```python
# Adding a title to the plot using the Matplotlib.

# draw barplot
sns.barplot(x="day", y="tip", data=tips)

## setting the title using Matplotlib
plt.title('Tip Amounts at Lunch and Dinner')

plt.show()
```

```python
# Setting Axis limits

# draw barplot
sns.barplot(x="day", y="tip", data=tips)

# setting the title using Matplotlib
plt.title('Tip Amounts at Lunch and Dinner')

## set the y limit of the plot
plt.ylim(1)

plt.show()
```

```python
# Changing the theme

# draw barplot
sns.barplot(x="day", y="tip", data=tips)

# setting the title using Matplotlib
plt.title('Tip Amounts at Lunch and Dinner')

# set the y limit of the plot
plt.ylim(1)

## changing the theme to darkgrid
sns.set_style("darkgrid")

# note you have access to the following styles {darkgrid, whitegrid, dark, white, ticks}

plt.show()
```

```python
# Removing Spines

# draw barplot
sns.barplot(x="day", y="tip", data=tips)

# setting the title using Matplotlib
plt.title('Tip Amounts at Lunch and Dinner')

# set the y limit of the plot
plt.ylim(1)

# changing the theme to darkgrid
sns.set_style("ticks")

# note you have access to the following styles {darkgrid, whitegrid, dark, white, ticks}

## Removing the spines
sns.despine(bottom = True)

# sns.despine(bottom = True)

plt.show()
```

```python
# Scaling the plots

# change the figure size
plt.figure(figsize = (10, 4))

# draw barplot
sns.barplot(x="day", y="tip", data=tips)


# setting the title using Matplotlib
plt.title('Tip Amounts at Lunch and Dinner')

# set the y limit of the plot
plt.ylim(1)

# changing the theme to darkgrid
sns.set_style("ticks")

# note you have access to the following styles {darkgrid, whitegrid, dark, white, ticks}

# Removing the spines
sns.despine(bottom = True)
# sns.despine(bottom = True)

## Setting the scale of the plot
sns.set_context("poster", font_scale=1)

#other contexts are “paper”, “talk”, and “poster”. font_scale sets the font size.

plt.show()
```

# Import Data for Visualization

Our data will be taken from http://insideairbnb.com/get-the-data/. We will download and import the following data sets for Boston, Massachusetts, United States:
1. Detailed Listings data- [listings.csv.gz](http://data.insideairbnb.com/united-states/ma/boston/2022-06-13/data/listings.csv.gz)
2.Detailed Review Data - [reviews.csv.gz](http://data.insideairbnb.com/united-states/ma/boston/2022-06-13/data/reviews.csv.gz)

Note you will have to extract the .gz files to get the underlying .csv files that we will use for our analysis

```python
import pandas as pd
#Import the required Datasets
listings_df=pd.read_csv('/content/listings.csv')
reviews_df=pd.read_csv('/content/reviews.csv')

```

# Data Preprocessing


```python
# Let's see what the listings dataframe contains
listings_df.info()
```

```python
# Let's reduce the number of columns in the listing data frame by selecting a few columns
listings_df = listings_df[['id','host_since', 'name', 'description','neighbourhood_cleansed',
                         'room_type','review_scores_rating', 'review_scores_accuracy',
                         'review_scores_cleanliness', 'review_scores_checkin',
                         'review_scores_communication',
                         'review_scores_location','review_scores_value',
                         'host_identity_verified', 'number_of_reviews',
                         'reviews_per_month','price','latitude','longitude',
                         'host_is_superhost','availability_365']]
listings_df.head(5)
```

```python
# Display the columns in the listing dataframe
listings_df.info()
```

```python
reviews_df.head(5)
```

```python
#let's also rename the id column that was originally the id column for the reviews dataframe to review_id
listings_df=listings_df.rename(columns={"neighbourhood_cleansed":"neighborhood"})
listings_df.info()
```

```python
#let's also rename the id column that was originally the id column for the reviews dataframe to review_id
reviews_df=reviews_df.rename(columns={"id":"reviews_id"})
reviews_df.info()
```

```python
#We would like to work with the price column but it is not formatted as a float
listings_df['price']=listings_df['price'].str.replace('$','')
listings_df['price']=listings_df['price'].str.replace(',','')
listings_df['price']=listings_df['price'].astype(float)
listings_df.info()
```

```python
# Let's loook at the columns in the reviews dataframe
reviews_df.head(5)
```

```python
# get more info about reviews
reviews_df.info()
```

```python
# First extract the year and month from the date column
# This ensures the dates are in a format pandas can extract time components from
reviews_df['date'] = reviews_df['date'].apply(lambda x: pd.to_datetime(x))
reviews_df['year'] = pd.DatetimeIndex(reviews_df['date']).year
reviews_df['month'] = pd.DatetimeIndex(reviews_df['date']).month


reviews_df.info()
```

```python
# join the listing and review datasets
listings_df=listings_df.rename(columns={"id":"listing_id"})
listingReviews_df = pd.merge(listings_df,reviews_df, how='left', on='listing_id')

```

## Boxenplot

A boxen plot is an alternative graphic for displaying distribution and summary statistics that generates more quantiles than a standard box plot. These additional quantiles allow for more distributional specificity particularly when you have many outliers.

```python
# Let's show the price distribution for different room types using a boxenplot
# boxenplot
sns.catplot(data=listings_df, x='number_of_reviews', y='room_type', kind='boxen')
```

### Violin Plot

The violin plot is a combination of a box plot and a kernel density plot that displays data peaks.

Violin plots are utilized to evaluate the distribution of numerical data, and are particularly useful for comparing the distributions of multiple groups.

```python
reviewscores = listingReviews_df[['year', 'review_scores_rating', 'review_scores_accuracy', 'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 'review_scores_location']]
reviewscores=reviewscores.groupby(['year']).mean()
reviewscores
```

```python
# Let's create a plot that show the concerntation of review scores across years
sns.heatmap(reviewscores)
```

```python
# Including annotations and a different colormap in the heatmap
sns.heatmap(reviewscores, annot=True,  cmap="YlGnBu")
```

```python
#Let's find the correlations between the review scores

#creata a list of columns that you wish to include
cols =['review_scores_rating', 'review_scores_accuracy', 'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 'review_scores_location']

#Create a numpy array that contains the correlations between the columns in the list
corrs = np.corrcoef(reviewscores[cols].values.T)

# Create a  heatmap and place the values from the array
sns.set(font_scale=1)
sns.set(rc={'figure.figsize':(7,7)})
hm=sns.heatmap(corrs, cbar = True, annot=True, square = True, fmt = '.2f',
              yticklabels = cols, xticklabels = cols).set_title('Review Score Correlations ')
```
