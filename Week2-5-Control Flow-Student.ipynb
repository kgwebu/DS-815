## Python Conditions and If statements


### If, Else, Elif


#### IF



```python
# Example:  IF Statement
balance = 5000
payment = 6000

if payment > balance:
  print("You have an overdraft")
```

####  Else




```python
# Example: Else Statement
balance = 15000
payment = 6000
if payment > balance:
  print("You have an overdraft")
else:
  print("You don't have an overdraft")
```

#### Elif




```python
#Example: elif
openingBalance = 7000
payment = 6000

if payment > balance:
  print("You have an overdraft")
elif payment < openingBalance:
  print("You have $", openingBalance - payment, "in your account")
else:
  print("You have $0 in your account")
```

### And

```python
# Example: and
price = 699
memory = 16
storage = 512

if price < 700 and memory > 8 and storage >100:
  print ("Great deal. Buy now!")
else:
  print("Too expensive")
```

### OR

```python
# Example: or
price = 699
memory = 16

if price < 700 or memory > 16:
  print ("Great deal. Buy now!")
else:
  print("Too expensive")
```

### Nested IF



```python
# Example: Nested if

x = 9

if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")

```

### Pass Statement

##  While Loops



```python
# Example: While Loop
year = 0
principal = 100000
interest = 0.08

while principal <= 2000000:
  print(year, principal)
  year += 1
  principal = round(principal + (principal * interest),2)


```

### The break statement




```python
#Example: Break
year = 0
principal = 100000
interest = 0.08


while principal <= 2000000:
  print(year, principal)
  if year >= 30:
    break
  year += 1
  principal = round(principal + (principal * interest),2)

print("Done!")
```

### The continue Statement


```python
# Example: Continue

temp = 31

while temp < 90:
  temp += 1
  if temp <= 55:
    print (temp, "Fall Temperature")
  elif temp > 55 and temp <65:
      continue
  else:
    print (temp, "Summer Temperature")

print("Loop End")
```

## For Loops


```python
# Example: For loop
stocks = ['AAPL','GOOG','IBM','FB','F','V', 'G', 'GE']
for i in stocks:
  print(i)
```

```python
# Example: For loop, in , not in
stocks = ['AAPL','GOOG','IBM','FB','F','V', 'G', 'GE']
selected = ['AAPL','IBM']
not_selected = []

for i in stocks:
  if i not in selected:
    not_selected.append(i)
print(not_selected)
```
