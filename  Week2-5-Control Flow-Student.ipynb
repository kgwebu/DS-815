### The continue Statement


```python
# Example: Continue
temp = 32

while temp < 90:
  temp += 1
  if temp <= 55:
    print(temp, "Fall Temperature")
  elif temp > 55 and temp < 65:
    continue
  else:
    print(temp, "Summer Temperature")


```

### The break Statement

```python

# Example: Break with lists
stocks = ['AAPL','GOOG','IBM','FB','F','V', 'G', 'GE']
for  i in stocks:
  print(i)
  if i == 'FB':
    break
```

### The continue Statement



```python
# Example: continue with lists
stocks = ['AAPL','GOOG','IBM','FB','F','V', 'G', 'GE']
for i in stocks:
  if i == 'FB':
    continue
  print(i)
```

### The range() Function

```python
# Example: Range Stop
for i in range(10):
  print(i)
```

```python
# Example:  Range Start-Stop
for i in range(1, 100,3):
  print(i)
```

```python
# Example:  Range Start-Stop-Step
for i in range (0, 101, 2):
  print(i)
```

```python
# Example 1:  Range
stocks = ['AAPL','GOOG','IBM','FB','F','V', 'G', 'GE']
for i in range(len(stocks)):
  print(i)
```

```python
# Example 2:  Range
stocks = ['AAPL','GOOG','IBM','FB','F','V', 'G', 'GE']
for i in range(len(stocks)):
  print(i, stocks[i])
```

## Nested Loops



```python
# Example: Nested loop
adjective = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for i in adjective:
  for j in fruits:
    print(i, j)
```
