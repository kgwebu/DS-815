## Escape Characters
Escape characters permit you to insert illegal characters into a string.

You can use a backslash \  followed by the character you wish to use. Look at the following examples.

```python
# Example of an illegal character: a double quote inside a string that is surrounded by double quotes
quote = "Who said "Never argue with stupid people, they will drag you down to their level and beat you with experience.""
print(quote) # this produces a syntax error due to multiple quotes being used

```

```python
# Example: You can use an escape sequence to handle the quotes e.g.
quote = "Who said \"Never argue with stupid people, they will drag you down to their level and beat you with experience.\""
print(quote) # this produces an error due to multiple quotes being used
```
