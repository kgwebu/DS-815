###  Multiple Variables


#### One Value to Multiple Variables

```python
# Example: Single value to multiple variables
bestUniversity = topUniversity = newEnglandUniversity = "University of New Hampshire"
print(bestUniversity)
print(topUniversity)
print(newEnglandUniversity)
```

#### Many Values to Multiple Variables

```python
# Example: Multiple values to multiple variables
street, city, state, zip = "10 Garrison Ave", "Durham", "NH", "03824"
print(street)
print(city)
print(state)
print(zip)
```

### Unpack a Collection

```python
# Example: Unpacking list
address = ["10 Garrison Ave", "Durham", "NH", "03824"]
street, city, state, zip = address
print(street)
print(city)
print(state)
print(zip)
```

## Global Variables

.

```python
# Example of a Global Variable
chant = "On to Victory" # this is a global variable

def myfunc():
    print("Wildcats! " + chant)

myfunc()

print(chant)
```

```python
# Example of a Local Variable

#chant = ""  # this is a global variable

def myfunc():
    chant_local = "On to Victory"        # this is variable is inside the function
    print("Wildcats! " + chant_local)

myfunc()

print("Wildcats! " + chant_local)
```

Note that in the example above an error is produced becuase the name chant1 is a local variable declared outside the function and it is not available for use outside the function.
