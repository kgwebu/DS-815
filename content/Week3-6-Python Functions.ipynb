A Function is a block of code that may be used repeatedly to perform a single task.
Parameters can be pased to the function

## Creating a function


```python
# Example: Create a function

def greet():
    print("Hello I am a function")

```

## Calling a function

```python
# Example: Calling a function
greet()
```

## Arguments/Parameters

```python

# Example: Arguments/Parameters


def greet(firstName):
    print("Hello " + firstName)

greet("John")
```

```python
# Example: Passing Multiple Arguments

def greet(firstName, lastName = ""):
    print("Hello " + firstName + lastName)

greet("John", " Brown")

greet("Stacy")
```

```python
# Example: Create a Default Argument

def greet(firstName, lastName =" Smith"):
    print("Hello " + firstName + lastName)

greet("John", " Brown")

greet("Stacy")
```

```python
# Example: Handle Multiple Arguments

def players_position (player, position):
    print(player, "is a ", position)

players_position(player = "Salah", position = "Striker")
```

```python
# Example: **kargs
def player_position(**kargs):
    print(kargs)

player_position(player1 = "Salah", position1 = "Striker" , player2 = "Mantip", position2 = "Defender")
```

## Passing a List as an Argument

```python
# Example: pass list of arguments

breakfast = ["*Toast", "*Eggs", "*Cereal"]
def breakfast_menu(breakfast):
    for x in breakfast:
        print(x)

breakfast_menu(breakfast)
```

## Return Values

```python
# Example: Return Values
def simpleInterest(principal, rate, time):
    finalAmount = principal * (1 + rate*time)
    return finalAmount

simpleInterest(1000, .0675, 5)
```
