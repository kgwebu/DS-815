### Creating Arrays With a Defined Data Type




```python
#  Create an array with data type integer:
integers = np.array([1.36, 7.214, 3.149], dtype = 'I')
print(integers.dtype)


# Create an array with data type string:
integers = np.array([2021, 2022, 2023], dtype = 'S')
print(integers.dtype)
```

### Converting Data Type on Existing Arrays



```python
# Change data type from float to integer by using 'I' as parameter value:
floats = np.array([1.36, 7.214, 3.149])
floatToIntger = floats.astype('I')
print(floatToIntger.dtype)
```

```python
# Change data type from integer to string by using 'S' as parameter value:
integers = np.array([2021, 2022, 2023])
intgerstoStrings = integers.astype('S')
print(intgerstoStrings.dtype)
```

## Array Copy vs. View



```python
# Make a copy, change the original array, and display both arrays:
cars = np.array(["Toyota", "Camry", "2022"])
carsCopy = cars.copy()
print(carsCopy)
print(cars[1])

# Note The copy IS NOT  affected by the changes made to the original array.

```

```python
# Create a view, change the original array, and display both arrays:
cars = np.array(["Toyota", "Camry", "2022"])
carsView = cars.view()
print(carsView)
cars[1] = "Highlander"
print(cars[1])

# Note the  View IS  affected by the changes made to the original array.
```

### Check if Array Owns its Data


```python
#Display the value of the base attribute to check if an array owns it's data or not:
cars = np.array(["Toyota", "Camry", "2022"])
carsCopy = cars.copy()
carsView = cars.view()

print(carsCopy.base)
print(carsView.base)
```

## Array Join



### Joining Arrays using concatination




numpy.png

```python
# Join two 1-D arrays along row (axis=0):
letters1 = np.array(["A","B","C"])

letters2 = np.array(["D","E","F"])

letersJoined = np.concatenate((letters1, letters2), axis = 0)
print(letters1)
print(letters2)
print(letersJoined)
```

```python
#Join two 2-D arrays along columns[default] (axis=0):

letters1 = np.array([["A","B","C"],
                     ["D","E","F"]
                     ])

letters2 = np.array([["G","H","I"],
                     ["J","K","L"]
                     ])

letterJoined = np.concatenate((letters1, letters2), axis = 0)
print(letterJoined)
```

```python
# Join two 2-D arrays along rows (axis=1):

letters1 = np.array([["A","B","C"],
                     ["D","E","F"]
                     ])

letters2 = np.array([["G","H","I"],
                     ["J","K","L"]
                     ])

letterJoined = np.concatenate((letters1, letters2), axis = 1)
print(letterJoined)
```

### Joining Arrays Using Stack Functions


```python
# Stack two 1-D arrays:
letters1 = np.array(["A","B","C"])

letters2 = np.array(["D","E","F"])


letterJoined = np.stack((letters1, letters2))
print(letterJoined)
```

<b> NumPy provides a helper functions:</b>
* hstack() to stack along rows.
* vstack()  to stack along columns.
    

```python
# Example: vstack

letters1 = np.array(["A","B","C"])

letters2 = np.array(["D","E","F"])

letterJoined = np.vstack((letters1, letters2))
print(letterJoined)
```

```python
# Example: hstack
letters1 = np.array(["A","B","C"])

letters2 = np.array(["D","E","F"])

letterJoined = np.hstack((letters1, letters2))
print(letterJoined)
```

```python
# NumPy provides a helper function: dstack() to stack along height, which is the same as depth.
# Example: dstack

letters1 = np.array(["A","B","C"])

letters2 = np.array(["D","E","F"])

letterJoined = np.dstack((letters1, letters2))
print(letterJoined)
```

##  Filter Array


```python
# Example: Filter array based on one condition
numArray = np.array([1, 4, 2, 7, 9, 3, 5, 8])
arryFiltered = numArray [np.where(numArray > 5)]
print(arryFiltered )
```

```python
# Example: Filter array based on multiple conditions
numArray = np.array([1, 4, 2, 7, 9, 3, 5, 8])
arryFiltered = numArray [np.where((numArray > 5) & (numArray < 9))]
print(arryFiltered )
```
