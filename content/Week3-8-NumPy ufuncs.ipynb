## Simple Arithmetic functions



```python
# pip install numpy
import numpy as np
```

```python
# Example: Add
arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([30, 70, 90, 80, 20, 33])

newarr = np.add(arr1, arr2)

print(newarr)
```

```python
# Example: Subtract
arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([30, 7, 2, 8, 20, 33])

newarr = np.subtract(arr1, arr2)

print(newarr)
```

```python
# Example: Multiply
arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([30, 7, 2, 8, 20, 33])

newarr = np.multiply(arr1, arr2)

print(newarr)
```

```python
# Example: Divide
arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([5, 4, 2, 8, 5, 20])

newarr = np.divide(arr1, arr2)

print(newarr)
```

```python
# Example: Power
arr1 = np.array([5, 7, 9, 8, 3, 10])
arr2 = np.array([5, 4, 2, 6, 5, 3])

newarr = np.power(arr1, arr2)

print(newarr)
```

```python
#Example: Remainder

arr1 = np.array([5, 7, 9, 8, 3, 10])
arr2 = np.array([5, 4, 2, 6, 5, 3])

newarr = np.mod(arr1, arr2)

print(newarr)
```

```python
#Example: Remainder

arr1 = np.array([5, 7, 9, 8, 3, 10])
arr2 = np.array([5, 4, 2, 6, 5, 3])

```

```python
#Quotient and remainder

arr1 = np.array([5, 7, 9, 8, 3, 10])
arr2 = np.array([5, 4, 2, 6, 5, 3])
newarr = np.divmod(arr1, arr2)

print(newarr)
```

```python
#Absolute Values
arr1 = np.array([5, -7, 9, -8, -3, 10])

newarr = np.absolute(arr1)

print(newarr)
```

## Rounding Decimals



```python
#Truncate

arr1 = np.array([5.12312, 7.3453, 9.6343, 8.7454, 3.12324, 10.2])

newarr = np.trunc(arr1)

print(newarr)
```

```python
#Round

arr1 = np.array([5.12312, 7.3453, 9.6343, 8.7454, 3.12324, 10.2])
newarr = np.around(arr1)

print(newarr)
```

```python
#floor

arr1 = np.array([5.12312, 7.3453, 9.6343, 8.7454, 3.12324, 10.2])

newarr = np.floor(arr1)

print(newarr)
```

```python
#Ceil

arr1 = np.array([5.12312, 7.3453, 9.6343, 8.7454, 3.12324, 10.2])

newarr = np.ceil(arr1)

print(newarr)
```

## Summations



```python
# Add
arr1 = np.array([10, 20, 30])
arr2 = np.array([90, 80, 70])

newarr = np.add(arr1, arr2)

print(newarr)
```

```python
#Summation
arr1 = np.array([10, 20, 30])
arr2 = np.array([90, 80, 70])

newarr = np.sum([arr1, arr2])

print(newarr)
```

```python
# Summation over an axis

arr1 = np.array([10, 20, 30])
arr2 = np.array([90, 80, 70])

newarr = np.sum([arr1, arr2], axis=1)

print(newarr)
```

```python
# Cummulative Sum
arr1 = np.array([10, 20, 30])
newarr = np.cumsum(arr1)

print(newarr)
```

 ## Products




```python
#Multiply 2 arrays
arr1 = np.array([3, 4, 3])
arr2 = np.array([3, 2, 4 ])

newarr = np.prod([arr1, arr2])

print(newarr)
```

```python
#Product 1 array

arr1 = np.array([10, 20, 30])

newarr = np.prod(arr1)

print(newarr)
```

```python
# Product 2 arrays

arr1 = np.array([3, 4, 3])
arr2 = np.array([3, 2, 4 ])
newarr = np.prod([arr1, arr2], axis=1)

print(newarr)
```

```python
# Product 2 arrays

arr1 = np.array([3, 4, 3])
arr2 = np.array([3, 2, 4 ])

```

```python
# Cummulative product

arr1 = np.array([10, 20, 30])

newarr = np.cumprod(arr1)

print(newarr)
```

## Differences


```python
#Product 1 array

arr1 = np.array([10, 20, 59])

newarr = np.diff(arr1)

print(newarr)
```

```python
# diff 2 arrays

arr1 = np.array([3, 4, 3])
arr2 = np.array([3, 2, 4 ])

newarr = np.diff([arr1, arr2])

print(newarr)
```
