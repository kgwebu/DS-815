###  Nested Dictionaries
A dictionary can contain dictionaries, this is called nested dictionaries.





```python
## Example of a nested Dictionary

companyTicker = {
    "Apple Inc": {
       "ticker": "AAPL",
        "SIC" : 3571
    },
    "Boeing Co":{
       "ticker": "BA",
        "SIC" : 3721
    },
    "Gap Inc": {
       "ticker": "GPS",
        "SIC" : 5651
    },
    "Pfizer Inc": {
       "ticker": "PFE",
        "SIC" : 5912
    },
    "Target Corp": {
       "ticker": "TGT",
        "SIC" : 5411
    },
    "Tiffany & Co": {
       "ticker": "TIF",
    }
}

print(companyTicker)
```
