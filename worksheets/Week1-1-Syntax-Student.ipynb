# Using Kaggle Notebooks  

Kaggle notebooks are like Jupyter notebooks in the cloud — everything is ready for you!  

---

####🛠 Key Features
- **Code cells** → write and run Python code  
- **Markdown cells** → add text, notes, or explanations  
- **Datasets** → access Kaggle’s library or upload your own (`/kaggle/input/`)  
- **Free hardware** → choose CPU, GPU, or TPU for faster processing  
- **Autosave & sharing** → your work is saved automatically, and you can share or keep private  

---

#### ▶ Running Your First Program
1. Click inside a **code cell**  
2. Type your code, e.g.:  

```python
print("Hello, Kaggle!")
```

3. Press Shift + Enter (or the ▶ button) to run it.

4. The output appears right below the cell.
